## Accelerated MRI

```python
dataset = load_from_disk(PATH / 'hf/fastmri')
dataset.set_format('numpy')

x_gt = dataset['val'][256:1280:256]['x']
to_pil(x_gt, pad=4)
```

```python
A = jnp.stack([make_mask(r=r, key=rng.split()) for r in (4, 8, 16, 32)])
y = unflatten(measure(A, flatten(x_gt)), 320, 320)
y = y + 1e-2 * rng.normal(y.shape)

x_zero = ifft2c(real2complex(y)).real
to_pil(x_zero, pad=4)
```

```python
x_infer_1 = sample(model, y, A, rng.split(), steps=64, maxiter=3)
to_pil(x_infer_1, pad=4)
```

```python
x_infer_2 = sample(model, y, A, rng.split(), steps=64, maxiter=3)
to_pil(x_infer_2, pad=4)
```

```python
grid = jnp.stack((
    jnp.broadcast_to(A * 4 - 2, x_gt.shape),
    x_zero,
    x_infer_1,
    x_infer_2,
    x_gt,
), axis=1)

to_pil(grid, pad=4, file='fastmri_accelerated.png')
```

## Crop

```python
x_gt = dataset['val'][1280:2560:256]['x']
to_pil(x_gt, pad=4)
```

```python
@inox.vmap
def crop(x):
    x = unflatten(x, 320, 320)
    y = x[..., 80:-80, 80:-80, :]
    y = flatten(y)
    return y

y = unflatten(crop(flatten(x_gt)), 160, 160)
to_pil(y, pad=80 + 4)
```

```python
x_infer_1 = sample_any(
    model,
    shape=flatten(x_gt).shape,
    A=crop,
    y=flatten(y),
    cov_y=1e-3 ** 2,
    key=rng.split(),
    steps=64,
    maxiter=3,
)

x_infer_1 = unflatten(x_infer_1, 320, 320)
to_pil(x_infer_1, pad=4)
```

```python
x_infer_2 = sample_any(
    model,
    shape=flatten(x_gt).shape,
    A=crop,
    y=flatten(y),
    cov_y=1e-3 ** 2,
    key=rng.split(),
    steps=64,
    maxiter=3,
)

x_infer_2 = unflatten(x_infer_2, 320, 320)
to_pil(x_infer_2, pad=4)
```

```python
grid = [
    [to_pil(x_gt, pad=4)],
    [to_pil(y, pad=80+4)],
    [to_pil(x_infer_1, pad=4)],
    [to_pil(x_infer_2, pad=4)],
]

collate(grid, file='fastmri_crop.png')
```
